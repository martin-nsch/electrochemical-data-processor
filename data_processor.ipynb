```python
def weight_calc(sample):
    df = pd.read_excel(loc+"\\weight.xlsx")
    if ('ine' in sample):
        if ('naced' in sample):
            af = df[df.eq("Furnaced").any(axis=1)]
            weight = float(af['weight'].values)
        else:    
            af = df[df.eq("Pristine").any(axis=1)]
            weight = float(af['weight'].values)

    elif (('cid' in sample) and ('naced' in sample)):
        if (('60') in sample):
            af = df[df.eq("Acid Etched 60min Furnaced").any(axis=1)]
            weight = float(af['weight'].values)
        elif ('45' in sample):
            af = df[df.eq("Acid Etched 45min Furnaced").any(axis=1)]
            weight = float(af['weight'].values)
        elif ('30' in sample):
            af = df[df.eq("Acid Etched 30min Furnaced").any(axis=1)]
            weight = float(af['weight'].values)
        elif('15' in sample):
            af = df[df.eq("Acid Etched 15min Furnaced").any(axis=1)]
            weight = float(af['weight'].values)
            
    elif ('naced' in sample):
        af = df[df.eq("Furnaced").any(axis=1)]
        weight = float(af['weight'].values)

    elif ('cid' in sample):
        if ('60' in sample):
            af = df[df.eq("Acid Etched 60min").any(axis=1)]
            weight = float(af['weight'].values)
        elif ('45' in sample):
            af = df[df.eq("Acid Etched 45min").any(axis=1)]
            weight = float(af['weight'].values)
        elif ('30' in sample):
            af = df[df.eq("Acid Etched 30min").any(axis=1)]
            weight = float(af['weight'].values)
        elif('15' in sample):
            af = df[df.eq("Acid Etched 15min").any(axis=1)]
            weight = float(af['weight'].values)
    else:
        weight = 0
    return(weight)
```
